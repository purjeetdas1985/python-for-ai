```python
## Function definitions
def profit_calculator(revenue, expenses):
    profit = revenue - expenses
    print(f"Profit: {profit}")
    return profit

profit_calculator(10000, 7000)
```

```python
## Functions returning more than one argument

def profit_calculator(revenue, expenses):
    profit = revenue - expenses
    margin = profit / revenue
    print(f"Profit: {profit}")
    return profit,margin

print(profit_calculator(10000, 7000))
```

```python
## Functions returning dictionaries

def profit_calculator(revenue, expenses):
    profit = revenue - expenses
    margin = profit / revenue
    print(f"Profit: {profit}")
    return {"profit": profit, "margin": margin}

print(profit_calculator(10000, 7000))
```

```python
## Functions with multiple arguments
def additions(*args):
   
    return sum(args)

print(additions(1, 2, 3, 4, 5))
```

```python
##  Function with keyword arguments
def greet(**kwargs):
    return f"{kwargs.get('greeting', 'Hello')}, {kwargs.get('name')}!"

print(greet(name="Alice"))
```
