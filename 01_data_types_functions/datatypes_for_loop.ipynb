```python
## Variables and data types
revenue=100
expense=80

profit= revenue-expense
print(f"profit is {profit}")
margin= profit/revenue
print(f"margin is {margin}")

```

```python
## lists and loops

revenue=[10,30,50,70]
expense=[5,10,20,30]

for i in range(len(revenue)):
    profit= revenue[i]-expense[i]
    print(f"profit is {profit}")
    margin= profit/revenue[i]
    print(f"margin is {margin}")
```

```python
## Dictonaries and loops

financials=[{"revenue":10, "expense":5},            {"revenue":30, "expense":10},
            {"revenue":50, "expense":20},            {"revenue":70, "expense":30}]
for fin in financials:
    profit= fin["revenue"]-fin["expense"]
    fin["profit"]=profit
    margin= profit/fin["revenue"]
    fin["margin"]=margin
    
print(financials)
```

```python
## Tuples and loops

financials=[(10,5), (30,10), (50,20), (70,30)]
for fin in financials:
    profit= fin[0]-fin[1]
    margin= profit/fin[0]
    print(f"profit is {profit}")
    print(f"margin is {margin}")
```
