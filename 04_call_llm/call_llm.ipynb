```python
from dotenv import load_dotenv
from openai import OpenAI
import os
load_dotenv()
```

```python
OPENROUTER_BASE_URL = "https://openrouter.ai/api/v1"
client = OpenAI(
    base_url="https://openrouter.ai/api/v1",
    api_key=os.getenv("OPENROUTER_API_KEY"),
)
ingredients = "chicken, garlic, olive oil, lemon, rosemary"
cusine_type = "Mediterranean"
content= f"""
    You are a chef if I asked you to build a recipe out of the ingradentis I have in my fridge. 
    I will give you a list of ingredients and the cusine i like and you will give me a recipe that I can make with those ingredients. 
    You will not give me any other information other than the recipe. 
    The ingredients are : {ingredients}
    Cusine type is : {cusine_type}
    The recipe should be in the following format: \n\nRecipe Name: \nIngredients: \nInstructions: \n\nIf you understand, please respond with the recipe.
    """
print(f"Requested content: {content}")
try:
    resp = client.chat.completions.create(
        model="openrouter/auto",
        messages=[{"role": "user", "content": content}]
    )
    if resp.choices and len(resp.choices) > 0 and resp.choices[0].message:
        print(resp.choices[0].message.content)
except Exception as e:
    print(f"Error: {e}")

```
